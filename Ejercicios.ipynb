```python
import numpy as np
import matplotlib.pyplot as plt

# --- Simulación de Datos Inventados ---
# Generamos 10 puntos de datos para el eje X
x = np.linspace(0, 10, 10) 
# Generamos datos para el eje Y basados en una función simple (ej. cuadrática + ruido)
y = 2 * x**2 - 5 * x + 10 + np.random.normal(0, 15, 10)

# Mostramos los primeros 5 pares de datos
print("Datos simulados (X, Y):")
for i in range(5):
    print(f"({x[i]:.2f}, {y[i]:.2f})")
```

```python
# Crea la figura
plt.figure(figsize=(10, 6))

# Grafica los datos (scatter plot)
plt.scatter(x, y, color='darkorange', marker='o', label='Puntos de Datos Simulados')

# Añade etiquetas y título
plt.title('Simulación de un Conjunto de Datos')
plt.xlabel('Variable Independiente (X)')
plt.ylabel('Variable Dependiente (Y)')
plt.legend()
plt.grid(True)

# ESTA LÍNEA GUARDA EL GRÁFICO EN LA CARPETA 'plato'
plt.savefig('plots/grafico_simulado.png') 

# Muestra el gráfico
plt.show()
```
